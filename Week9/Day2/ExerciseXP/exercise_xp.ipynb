```python
URL = "https://en.wikipedia.org/"
HEADERS = {"User-Agent": "Mozilla/5.0 (bootcamp exercise)"}

request = Request(URL, headers=HEADERS)

with urlopen(request) as response:
    wikipedia_page_content = response.read().decode("utf-8")

soup = BeautifulSoup(wikipedia_page_content, "html.parser")
```

### **🌟 Exercise 4: Checking for Page Title**

### **Instructions**

Write a Python program to check whether a page contains a title or not.


```python
from urllib.request import Request, urlopen

from bs4 import BeautifulSoup

HEADERS = {"User-Agent": "Mozilla/5.0 (bootcamp exercise)"}


def fetch_html(url: str) -> str:
    """Download a page and return its HTML as a string."""
    request = Request(url, headers=HEADERS)

    with urlopen(request) as response:
        return response.read().decode("utf-8")


def get_page_title(html: str) -> str | None:
    """Return the page title text, or None if there is no usable <title>."""
    soup = BeautifulSoup(html, "html.parser")

    if soup.title is None:
        return None

    title = soup.title.get_text(strip=True)
    # Treat an empty <title></title> as "no title" too.
    return title or None
```

```python
url = "https://en.wikipedia.org/"
title = get_page_title(fetch_html(url))

if title:
    print(f"The page has a title: {title}")
else:
    print("The page does not have a title.")
```

### **🌟 Exercise 5: Analyzing US-CERT Security Alerts**

### **Instructions**

Write a Python program to get the number of security alerts issued by US-CERT in the current year.

[Source](https://www.cisa.gov/news-events/cybersecurity-advisories?f%5B0%5D=advisory_type%3A93)


```python
from datetime import datetime

import requests
from bs4 import BeautifulSoup

BASE_URL = "https://www.cisa.gov/news-events/cybersecurity-advisories"
PARAMS = {"f[0]": "advisory_type:93"}  # 93 = "Alert" filter from the exercise link
HEADERS = {"User-Agent": "Mozilla/5.0 (DI-Bootcamp web scraping exercise)"}
CURRENT_YEAR = datetime.now().year
MAX_PAGES = 100
```

```python
def fetch_page(page: int) -> BeautifulSoup:
    """Download one listing page and return it parsed."""
    response = requests.get(
        BASE_URL,
        params={**PARAMS, "page": page},
        headers=HEADERS,
        timeout=15,
    )
    response.raise_for_status()

    return BeautifulSoup(response.text, "html.parser")


def parse_items(soup: BeautifulSoup) -> list[dict]:
    """Extract date, type label and title from each teaser on a page."""
    items = []
    for article in soup.select("article.c-teaser"):
        time_tag = article.find("time")
        meta_tag = article.select_one(".c-teaser__meta")
        title_tag = article.select_one(".c-teaser__title")
        if time_tag is None:
            continue

        items.append(
            {
                "date": datetime.fromisoformat(time_tag["datetime"][:10]).date(),
                "type": meta_tag.get_text(strip=True) if meta_tag else "",
                "title": title_tag.get_text(strip=True) if title_tag else "",
            }
        )

    return items
```

```python
def count_alerts_for_year(year: int) -> list[dict]:
    """Walk pages newest-first and collect Alerts from `year` until an older one appears."""
    alerts = []
    for page in range(MAX_PAGES):
        items = parse_items(fetch_page(page))
        if not items:
            break

        for item in items:
            if item["date"].year < year:
                return alerts

            if item["date"].year == year and item["type"].lower() == "alert":
                alerts.append(item)

    return alerts


alerts = count_alerts_for_year(CURRENT_YEAR)
print(f"US-CERT/CISA security alerts issued in {CURRENT_YEAR}: {len(alerts)}")
```
