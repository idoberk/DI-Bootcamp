**🌟 Exercise 2 : Type Conversion and Array Operations**

Instructions

Convert a list [3.14, 2.17, 0, 1, 2] into a NumPy array and convert its data type to integer.

Expected Output:

`array([3, 2, 0, 1, 2])`


```python
list_arr = [3.14, 2.17, 0, 1, 2]
ex_2_arr = np.array(list_arr)
ex_2_arr = ex_2_arr.astype(int)
ex_2_arr
```
