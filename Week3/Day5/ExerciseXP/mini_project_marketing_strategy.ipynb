```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_excel("US Superstore data.xls")
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Dropping irrelevant columns
df = df.drop(columns=["Row ID", "Country", "Postal Code"])
```

```python
df["State"].value_counts()
```

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
print(df[["Order Date"]].head())
print(df[["Order Year", "Order Month"]].head())
```

```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False)
state_sales.head(10)
```

```python
state_sales.head(10).sort_values().plot(kind="barh", figsize=(8, 6))
plt.xlabel("Total Sales ($)")
plt.title("Top 10 States by Sales")
plt.show()
```

```python
# Get total sales and profit for New York and California side by side.
ny_ca = df[df["State"].isin(["New York", "California"])]
comparison = ny_ca.groupby("State")[["Sales", "Profit"]].sum()
comparison
```

```python
comparison["Order Count"] = ny_ca.groupby("State").size()
comparison["Avg Profit Margin"] = comparison["Profit"] / comparison["Sales"]
comparison
```

```python
comparison[["Sales", "Profit"]].plot(kind="barh", figsize=(6, 5))
plt.title("New York vs. California: Sales & Profit")
plt.xlabel("Amount ($)")
plt.show()
```

```python
# Finding who is an outstanding customer in New York (What does outstanding means in this context? Most total sales? Highest total profit? Most order count?)

ny = df[df["State"] == "New York"]
ny_customers = (
    ny.groupby("Customer Name")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)
ny_customers.head(10)
```

```python
# Getting New York customers and checking their Sales, Profit and Order Count.
ny_customers["Order Count"] = ny.groupby("Customer Name").size()
top_ny = ny_customers.sort_values("Profit", ascending=False).head(10)
print(top_ny)

top_ny[["Sales", "Profit"]].plot(kind="bar", figsize=(10, 6))
plt.title("Top 10 New York Customers by Sales & Profit")
plt.xlabel("Customer")
plt.ylabel("Amount ($)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Are there any differences among states in profitability? What does it mean exactly? Difference in Profit? Profit Margin (Profit / Sales)?
state_profit = df.groupby("State")[["Sales", "Profit"]].sum()
state_profit["Profit Margin"] = state_profit["Profit"] / state_profit["Sales"]
state_profit.sort_values("Profit Margin", ascending=False)
```

```python
sorted_states_by_profit_margin = state_profit.sort_values("Profit Margin")[
    "Profit Margin"
]
colors = ["red" if x < 0 else "steelblue" for x in sorted_states_by_profit_margin]
sorted_states_by_profit_margin.plot(kind="barh", figsize=(8, 10), color=colors)
plt.xlabel("Profit Margin")
plt.title("Profit Margin by State")
plt.axvline(0, color="black")
plt.show()
```

```python
customer_profit = (
    df.groupby("Customer Name")["Profit"].sum().sort_values(ascending=False)
)

pareto_df = customer_profit.reset_index()
pareto_df["Cumulative Profit %"] = (
    pareto_df["Profit"].cumsum() / pareto_df["Profit"].sum() * 100
)
pareto_df["Customer %"] = (pareto_df.index + 1) / len(pareto_df) * 100

pareto_df.head(10)
```

```python
top_80 = pareto_df[pareto_df["Cumulative Profit %"] <= 80]
n_customers_for_80percent = len(top_80) + 1
percent_of_customers = n_customers_for_80percent / len(pareto_df) * 100
print(
    f"{n_customers_for_80percent} customers ({percent_of_customers:.1f}%) generate ~80% of total profit"
)
```

```python
plt.plot(pareto_df["Customer %"], pareto_df["Cumulative Profit %"])
plt.axhline(80, color="red", linestyle="--")
plt.axvline(20, color="green", linestyle="--")
plt.grid(alpha=0.3)
plt.xlabel("% of Customers")
plt.ylabel("Cumulative % of Profit")
plt.title("Pareto Analysis: Customers vs. Profit")
plt.show()
```

```python
# Checking if we have Identical city names in different states.
city_state_pairs = df[["City", "State"]].drop_duplicates()
duplicated_cities = city_state_pairs[city_state_pairs["City"].duplicated(keep=False)]
duplicated_cities.sort_values("City")
```

```python
city_stats = df.groupby(["City", "State"])[["Sales", "Profit"]].sum()
city_stats["Profit Margin"] = city_stats["Profit"] / city_stats["Sales"]

top_20_cities_by_sales = city_stats.sort_values("Sales", ascending=False).head(20)
top_20_cities_by_profit = city_stats.sort_values("Profit", ascending=False).head(20)

print(f"Top 20 cities by sales:\n {top_20_cities_by_sales}")
print(f"Top 20 cities by profit:\n {top_20_cities_by_profit}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 10))

top_20_cities_by_sales.reset_index().assign(
    Label=lambda d: d["City"] + ", " + d["State"]
).set_index("Label")["Sales"].sort_values().plot(kind="barh", ax=axes[0])

axes[0].set_xlabel("Total Sales ($)")
axes[0].set_title("Top 20 Cities by Sales")

top_20_cities_by_profit.reset_index().assign(
    Label=lambda d: d["City"] + ", " + d["State"]
).set_index("Label")["Profit"].sort_values().plot(
    kind="barh", ax=axes[1], color="orange"
)

axes[1].set_xlabel("Total Profit ($)")
axes[1].set_title("Top 20 Cities by Profit")

for ax in axes:
    ax.grid(axis="x", linestyle="--", alpha=0.85)
    ax.set_ylabel("City")

plt.tight_layout()
plt.show()
```

```python
df.groupby("Customer Name")["Customer ID"].nunique().max()
```

```python
top_20_customers_by_sales = (
    df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False)
).head(20)

top_20_customers_by_sales
```

```python
top_20_customers_by_sales.sort_values().plot(kind="barh", figsize=(8, 10))
plt.xlabel("Total Sales ($)")
plt.ylabel("Customer")
plt.title("Top 20 Customers by Sales")
plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
customer_sales = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False)

sales_pareto_df = customer_sales.reset_index()
sales_pareto_df["Cumulative Sales %"] = (
    sales_pareto_df["Sales"].cumsum() / sales_pareto_df["Sales"].sum() * 100
)
sales_pareto_df["Customer %"] = (sales_pareto_df.index + 1) / len(sales_pareto_df) * 100

sales_pareto_df.head(10)
```

```python
top_80_sales = sales_pareto_df[sales_pareto_df["Cumulative Sales %"] <= 80]
n_customers_for_80percent_sales = len(top_80_sales) + 1
percent_customers = n_customers_for_80percent_sales / len(sales_pareto_df) * 100
print(
    f"{n_customers_for_80percent_sales} customers ({percent_customers:.1f}%) generate ~80% of total sales"
)
```

```python
plt.plot(sales_pareto_df["Customer %"], sales_pareto_df["Cumulative Sales %"])
plt.axhline(80, color="red", linestyle="--")
plt.axvline(20, color="green", linestyle="--")
plt.xlabel("% of Customers")
plt.ylabel("Cumulative % of Sales")
plt.title("Pareto Analysis: Customers vs. Sales")
plt.show()
```

### Insights: Top States by Sales

California leads by a wide margin (**$457,688** in total sales), followed by New York (**$310,876**) and a steep drop to Texas (**$170,188**). The top 2 states alone account for a disproportionate share of total sales compared to the rest of the list - California and New York combined outsell the next 8 states shown.

---

### Insights: New York vs. California

Despite California generating ~47% more sales than New York (**$457,688** vs **$310,876**) and having nearly double the order count (**2,001** vs **1,128**), the two states produce almost identical total profit (**$76,381** vs **$74,039**). This reveals a big efficiency gap: New York's profit margin is **23.8%** versus California's **16.9%** - New York converts sales into profit about 43% more efficiently. Raw sales volume is a misleading indicator of a state's value here; margin tells a different story.

---

### Insights: Outstanding Customer in New York

**Tom Ashbrook** is the standout customer in New York: highest total sales (**$13,723**) and highest total profit (**$4,599**) among all NY customers, achieved in only **5 orders** - roughly $2,745 sales and $920 profit per order, well above every other customer on the list. He leads on every dimension (sales, profit, and efficiency), not just one.

---

### Insights: Profitability Differences Among States

Profit margin varies substantially by state, and several states are outright unprofitable overall. States like **District of Columbia (36.9%)**, **Delaware (36.3%)**, and **Minnesota (36.2%)** run high margins, while other states have **negative** average margins - meaning the company loses money overall on sales made there, despite generating revenue. Profitability is clearly not proportional to sales volume across states.

---

### Insights: Pareto Principle - Customers and Profit

The Pareto principle holds strongly here, even sharper than the classic 80/20 split: just **153 customers (19.3%)** of all 793 customers generate **80%** of total profit. Notably, the cumulative profit curve actually peaks _above_ 100% (reaching **~124.9%**) around customer #638, before settling back to 100% - because **155 customers (~19.5%)** have net-negative profit, actively eroding the gains made by the top performers.

This means the real story isn't just "focus on your best customers" - it's "a small group drives outsized profit, and a separate small group is actively costing the business money."

---

### Insights: Top 20 Cities by Sales and Profit

New York City, Los Angeles, Seattle, and San Francisco top **both** the sales and profit rankings, showing consistent strength on both fronts. But several high-sales cities are unprofitable: **Philadelphia** ($109,077 in sales) posts a **-$13,838 loss** (-12.7% margin), and Houston, Chicago, San Antonio, Dallas, and Jacksonville all show negative profit despite ranking in the top 20 by sales.

Conversely, cities like **Lafayette, IN** (45.7% margin) and **Atlanta** (40.7% margin) post strong profitability without cracking the top-15 by sales - high sales volume is not a reliable predictor of profitability at the city level.

> **Note:** City name collisions were an issue here - 57 city names (e.g. "Springfield," "Columbus") repeat across 39 different states, so grouping by `City` **and** `State` together was necessary to avoid merging unrelated cities.

---

### Insights: Top 20 Customers by Sales

**Sean Miller** is the top customer by sales (**$25,043**), ahead of Tamara Chand ($19,052) and Raymond Buch ($15,117). Several names from the profit-based Pareto analysis reappear here (Tamara Chand, Raymond Buch, Tom Ashbrook), suggesting some customers are valuable on both sales and profit - but Sean Miller's absence from the top profit contributors is worth flagging as a "high sales, not necessarily high profit" case.

---

### Insights: Cumulative Sales Curve and Pareto for Sales

Unlike profit, sales is far more evenly distributed across the customer base: it takes **396 customers (49.9%)** to reach 80% of total sales, compared to just **153 customers (19.3%)** for profit. The Pareto principle does **not** meaningfully apply to sales in this dataset - revenue is spread broadly, while profit is highly concentrated.

This is an important distinction: identifying "valuable" customers by sales volume alone would miss the real profit drivers.

---

### Conclusion: Marketing Strategy Priorities

Based on the analysis, prioritization should weight **profit efficiency** over raw sales volume:

- **States to prioritize:** New York (higher margin than California despite lower sales) and other high-margin states (DC, Delaware, Minnesota) over high-volume-but-lower-margin states.
- **Cities to prioritize:** New York City, Los Angeles, Seattle, and San Francisco (strong on both sales and profit), plus efficient smaller markets like Lafayette, IN and Atlanta. Deprioritize or investigate discount/pricing strategy in Philadelphia, Houston, Chicago, San Antonio, Dallas, and Jacksonville - all generate meaningful sales but currently lose money.
- **Customers to prioritize:** Focus retention/loyalty efforts on the ~19% of customers (like Tom Ashbrook, Tamara Chand) who drive 80% of profit, rather than the broader ~50% needed to reach 80% of sales - profit concentration, not sales volume, should guide customer-focused marketing spend.
- **Underlying gap to address:** ~155 customers and multiple cities/states have negative profit despite nonzero sales - likely tied to discounting patterns (worth a follow-up look at the `Discount` column), since this is actively working against overall profitability.
