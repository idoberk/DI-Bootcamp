### **Exercise 2: Descriptive Statistics**

- **Task:** Given a sample dataset, calculate the mean, median, variance, and standard deviation using `SciPy`.

**Sample Dataset:**

```
data = [12, 15, 13, 12, 18, 20, 22, 21]
```


````python
data = [12, 15, 13, 12, 18, 20, 22, 21]
mean = scipy.stats.tmean(data)
median = scipy.stats.scoreatpercentile(data, 50)
std = scipy.stats.tstd(data)
var = scipy.stats.tvar(data)

print(
    f"Mean: {mean:.2f}, Median: {median:.2f}, Standard Deviation: {std:.2f}, Variance: {var:.2f}"
)
````

### **Exercise 3: Understanding Distributions**

- **Task:** Generate a normal distribution using `SciPy` with a mean of 50 and a standard deviation of 10. Plot this distribution.

**Code Example:**

```
import numpy as np

```


````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
````

````python
mean, std = 50, 10
x = np.linspace(mean - 4 * std, mean + 4 * std, 500)
y = norm.pdf(x, mean, std)

plt.plot(x, y)
plt.title("Normal Distribution (mean = 50, standard deviation = 10)")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.show()
````

### **Exercise 4: T-Test Application**

- **Task:** Perform a T-test on two sets of randomly generated data.

**Code Example:**

```
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# complete your code here
```


````python
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

mean_data_1 = scipy.stats.tmean(data1)
mean_data_2 = scipy.stats.tmean(data2)

print(f"Mean data1: {mean_data_1:.2f}, Mean data2: {mean_data_2:.2f}")

std_data_1 = scipy.stats.tstd(data1)
std_data_2 = scipy.stats.tstd(data2)

print(
    f"Standard Deviation data1: {std_data_1:.2f}, Standard Deviation data2: {std_data_2:.2f}"
)

t_stat, p_val = scipy.stats.ttest_ind(data1, data2)

print(f"T-Stat: {t_stat}, P-Val: {p_val}")
````

### **Exercise 5: Linear Regression Analysis**

**Objective:** Apply linear regression to a dataset and interpret the results.

- **Task:** Given a dataset of housing prices (`house_prices`) and their corresponding sizes (`house_sizes`), use linear regression to predict the price of a house given its size.

**Dataset:**

- `house_sizes`: [50, 70, 80, 100, 120] (in square meters)
- `house_prices`: [150,000, 200,000, 210,000, 250,000, 280,000] (in currency units)

**Questions:**

- What is the slope and intercept of the regression line?
- Predict the price of a house that is 90 square meters.
- Interpret the meaning of the slope in the context of housing prices.


````python
house_sizes = np.array([50, 70, 80, 100, 120])  # In square meters
house_prices = np.array([150000, 200000, 210000, 250000, 280000])  # In currency units

slope, intercept, r_value, p_value, std_err = scipy.stats.linregress(
    house_sizes, house_prices
)

print(f"Slope: {slope:.2f}, Intercept: {intercept:.2f}")

predicted_price = slope * 90 + intercept

print(f"Predicted price for a 90 sqm house: {predicted_price:.2f}")
````

With house size as the predictor (independent variable = x) and house price as the price (dependent variable = y), the slope is about 1828 currency units per sqm.

The slope is the rate of change. For every additional sqm of house size, the predicted price increases by roughly 1828 currency units.

A positive slope like this confirms the expected relationship between price and size. The bigger houses cost more.


### **Exercise 6: Understanding ANOVA**

**Objective:** Test understanding of ANOVA and its application.

- **Task:** Three different fertilizers are applied to three separate groups of plants to test their effectiveness. The growth in centimeters is recorded.

**Dataset:**

- `fertilizer_1`: [5, 6, 7, 6, 5]
- `fertilizer_2`: [7, 8, 7, 9, 8]
- `fertilizer_3`: [4, 5, 4, 3, 4]

**Questions:**

- Perform an ANOVA test on the given data. What are the F-value and P-value?
- Based on the P-value, do the fertilizers have significantly different effects on plant growth?
- Explain what would happen if the P-value were greater than 0.05.


````python
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = scipy.stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"F-Value: {f_value}, P-Value: {p_value}")
````

Yes, the fertilizers have significantly different effects on plant growth. P-Value of 0.0000275 is far below the standard 0.05, so the fertilizers don't produce the same mean growth.

If P-Value > 0.05 then we wouldn't have enough evidence to conclude that the effects of the fertilizers are different. Any differences seen in the sample means could be due to random variation alone, and we couldn't confidently recommend one fertilizer of another based on this data.


### **Exercise 7: Probability Distributions (Optional)**

- Work with a binomial distribution: calculate probabilities for different numbers of successes.
- Example: Calculating the probability of getting exactly 5 heads in 10 coin flips.


````python
n, p = 10, 0.5
k_values = np.arange(0, n + 1)
probabilities = scipy.stats.binom.pmf(k_values, n, p)

for k, prob in zip(k_values, probabilities):
    print(f"P(exactly {k} heads): {prob:.4f}")

plt.bar(k_values, probabilities)
plt.title(f"Binomial Distribution (n={n}, p={p})")
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.xticks(k_values)
plt.show()
````

### **Exercise 8: Correlation Coefficients (Optional)**

- Calculate the Pearson and Spearman correlation coefficients between two variables in a dataset.
- Example in code: `data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})`


````python
import pandas as pd
````

````python
data = pd.DataFrame(
    {"age": [23, 25, 30, 35, 40], "income": [35000, 40000, 50000, 60000, 70000]}
)

pearson_corr, _ = scipy.stats.pearsonr(data["age"], data["income"])
spearman_corr, _ = scipy.stats.spearmanr(data["age"], data["income"])

print(f"Pearson Correlation Coefficient: {pearson_corr}")
print(f"Spearman Rank Correlation: {spearman_corr}")

plt.scatter(data["age"], data["income"], label="Data")

slope, intercept, _, _, _ = scipy.stats.linregress(data["age"], data["income"])
line_x = np.linspace(data["age"].min(), data["age"].max(), 100)
plt.plot(line_x, slope * line_x + intercept, color="red", label="Linear fit")

plt.title(
    f"Age vs Income (Pearson r={pearson_corr:.4f}, Spearman r={spearman_corr:.4f})"
)
plt.xlabel("Age")
plt.ylabel("Income")
plt.legend()
plt.show()
````
